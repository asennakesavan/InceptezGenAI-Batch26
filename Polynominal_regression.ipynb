```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_percentage_error
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/Laxminarayen/InceptezGenAI-Batch26/refs/heads/main/21.%20Polynomial%20Regression/handson/Salary_Experience_Dataset.csv")
```

```python
df.head(5)
```

```python
scaler = MinMaxScaler()
df['edu_norm'] = scaler.fit_transform(df[['years_experience']])

Lin_model = LinearRegression() # Instantiate the model
Lin_model.fit(df[['years_experience']], df['salary_lpa']) # Pass X as a DataFrame and y as a Series
print(f"   Salary = {Lin_model.intercept_:,.0f} + {Lin_model.coef_[0]:,.0f} \u00d7 years_experience")

Lin_model1 = LinearRegression() # Instantiate the model
Lin_model1.fit(df[['edu_norm']], df['salary_lpa']) # Pass X as a DataFrame and y as a Series
print(f"   Salary = {Lin_model1.intercept_:,.0f} + {Lin_model1.coef_[0]:,.0f} \u00d7 edu_norm")
```

```python
df.head(5)
```

create a model for polynominal regrssion.

```python
df['YOE_degree_2'] = df['years_experience']**2
```

```python
df['YOE_degree_3'] = df['years_experience']**3
```

```python
df.head(5)
```

```python
model_feature_cols = ['years_experience','education_level','certifications','edu_norm','YOE_degree_2','YOE_degree_3']
Scaler1 = MinMaxScaler()
normalized = Scaler1.fit_transform(df[model_feature_cols])
for index,feature in enumerate(model_feature_cols):
  df[feature + '_norm1'] = normalized[:,index]
df[[c + '_norm1' for c in model_feature_cols]].describe().loc[['min','max']]
df.drop(columns='edu_norm_norm1')


```

```python
train_df,test_df = train_test_split(df,test_size=0.2,random_state=1)
train_df =train_df.reset_index(drop=True)
test_df =test_df.reset_index(drop=True)
print('train_df ',train_df.shape)
print('test_df ',test_df.shape)
```

```python
model_a_feature = ['years_experience_norm1','education_level_norm1','certifications_norm1']
model_a = LinearRegression()
model_a.fit(train_df[model_a_feature],train_df['salary_lpa'])
for feature,coef in zip(model_a_feature,model_a.coef_):
  print(f"{feature:<}  is {coef:>}")
print(f"Intercept  is {model_a.intercept_}")
model_a_predict = model_a.predict(test_df[model_a_feature])
print(f" r2 value is {r2_score(test_df['salary_lpa'], model_a_predict):.2f}")
print("Predicted:", model_a_predict[:10])
print("Actual:", test_df['salary_lpa'].values[:10])
```

```python
model_b_feature = ['years_experience_norm1','education_level_norm1','certifications_norm1','YOE_degree_2_norm1']
model_b = LinearRegression()
model_b.fit(train_df[model_b_feature],train_df['salary_lpa'])
for feature,coef in zip(model_b_feature,model_b.coef_):
  print(f"{feature:<} is {coef:>}")
print(f"Intercept is {model_b.intercept_}")
model_b_predict = model_b.predict(test_df[model_b_feature])
print(f"r2 value is {r2_score(test_df['salary_lpa'],model_b_predict)}")
print(model_b_predict[:10])
```

```python
model_c_feature = ['years_experience_norm1','education_level_norm1','certifications_norm1','YOE_degree_2_norm1','YOE_degree_3_norm1']
model_c = LinearRegression()
model_c.fit(train_df[model_c_feature],train_df['salary_lpa'])
for feature,coef in zip(model_c_feature,model_c.coef_):
  print(f"{feature:<} is {coef:>}")
print(f"Intercept is {model_c.intercept_}")
```
